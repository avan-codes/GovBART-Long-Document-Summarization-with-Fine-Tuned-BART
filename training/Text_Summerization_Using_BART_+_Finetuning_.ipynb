```python
input = tokenizer(
    text, return_tensors="pt",
    truncation=True
)

summery_ids = model.generate(
    input['input_ids'] ,
    # max_length=80,
    # min_length=20,
    num_beams=4
)

summery = tokenizer.decode(summery_ids[0],
                           skip_special_tokens=True
                           )

```

##Preprocessing

```python
def preprocess(batch):
    source = batch['report']
    target = batch['summary']
    source_ids = tokenizer(source, truncation=True, padding='max_length', max_length=128)
    target_ids = tokenizer(target, truncation=True, padding='max_length', max_length=128)
    labels = target_ids['input_ids']
    labels = [(label if label != tokenizer.pad_token_id else -100)
    for label in labels]

    return {
        'input_ids':source_ids['input_ids'],
        'attention_mask':source_ids['attention_mask'],
        'labels':labels
    }
```

```python
df = ds.map(preprocess, batched=True)
```

```python
import torch
device = "cuda" if torch.cuda.is_available() else "cpu"
model.to(device)
```

```python
#@title Training Arguments
from transformers import TrainingArguments, Trainer

training_args = TrainingArguments(
    output_dir = '/content/drive/MyDrive/30_days_project_streak/day_1',
    per_device_train_batch_size=8,
    num_train_epochs=4)
```

```python
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=df['train'],
    eval_dataset=df['test']
)
```

```python
trainer.train()
```
